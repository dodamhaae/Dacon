# 
## <span style='color:#0000FF; font-family:cursive'> Air Conditioning </span>

( 냉방온도 + 난방온도 ) / 2 = 기준온도

```python
air = df[['냉방온도', '난방온도', '기준온도']]
air.head()
```

### 
#### DataFrame에서 결측치 확인

```python
air.isna().sum().to_frame(name='NaN')
```

```python
# 기준온도 1행을 제외하고, 모두 같은 index에서만 결측치 존재
len(air.loc[air['냉방온도'].isna() & air['난방온도'].isna() & air['기준온도'].isna()])
```

### 
#### Graph에서 결측치 확인

```python
def custom_legend_name(new_names):
    for i, new_name in enumerate(new_names):
        fig.data[i].name = new_name
```

```python
fig = make_subplots(rows=3, cols=1, shared_xaxes=True)

for i, c in enumerate(air.columns, start=1):
    fig.add_trace(go.Scatter(x=df['시간'], y=air[c]), row=i, col=1)
custom_legend_name(air.columns)
fig.show()
```

### 
### 1. 기준온도 결측치 보간 ➡️ 냉/난방온도가 존재하는 경우

내부온도관측치 - 외부온도관측치 = 내외부온도차  
그 결과, 결측치가 **685** 에서 **684** 로 줄었다.  
⠀ →⠀ **_1_** 개의 결측치 채워짐

```python
air['기준온도'] = np.where(air['기준온도'].isna(), (air['냉방온도']+air['난방온도'])/2, air['기준온도'])
air.isna().sum().to_frame(name='NaN')
```

### 2. 냉/난방온도 결측치 보간 ➡️ interpolate 함수

```python
air['냉방온도'] = air['냉방온도'].interpolate()
air['난방온도'] = air['난방온도'].interpolate()
```

### 
냉방온도와 난방온도의 결측치가 모두 채워짐을 알 수 있다.

```python
air.isna().sum().to_frame(name='NaN')
```

### 
### 3. 기준온도 결측치 보간

```python
air['기준온도'] = np.where(air['기준온도'].isna(), (air['냉방온도']+air['난방온도'])/2, air['기준온도'])
air.isna().sum().to_frame(name='NaN')
```

### 
모든 결측치가 채워짐을 알 수 있다.

```python
fig = make_subplots(rows=3, cols=1, shared_xaxes=True)

for i, c in enumerate(air.columns, start=1):
    fig.add_trace(go.Scatter(x=df['시간'], y=air[c]), row=i, col=1)
custom_legend_name(air.columns)
fig.show()
```

# 
## <span style='color:#0000FF; font-family:cursive'> Light </span>

백색광추정광량 + 적색광추정광량 + 청색광추정광량 = 총추정광량

```python
light = df[['백색광추정광량', '적색광추정광량', '청색광추정광량', '총추정광량']]
light.head()
```

### 
#### DataFrame에서 결측치 확인

```python
light.isna().sum().to_frame(name='NaN')
```

```python
# 청색광추정광량 1행을 제외하고, 모두 같은 index에서만 결측치 존재
len(light.loc[light['백색광추정광량'].isna() & light['적색광추정광량'].isna()
              & light['청색광추정광량'].isna()] & light['총추정광량'].isna())
```

### 
#### Graph에서 결측치 확인

```python
fig = make_subplots(rows=4, cols=1, shared_xaxes=True)

for i, c in enumerate(light.columns, start=1):
    fig.add_trace(go.Scatter(x=df['시간'], y=light[c]), row=i, col=1)
custom_legend_name(light.columns)
fig.show()
```

### 
### 1. 청색광추정광량 결측치 보간 ➡️ 백색광/적색광/총추정광량이 존재하는 경우

청생광추정광량 = 총추정광량 - ( 백색광추정광량 + 적색광추적광량 )  
그 결과, 결측치가 **690** 에서 **689** 로 줄었다.  
⠀ →⠀ **_1_** 개의 결측치 채워짐

```python
light['청색광추정광량'] = np.where(light['청색광추정광량'].isna(),
                          light['총추정광량']-(light['백색광추정광량']+light['적색광추정광량']), light['청색광추정광량'])
light.isna().sum().to_frame(name='NaN')
```

### 2. 백/적/청색광추정광량 결측치 보간 ➡️ interpolate 함수

```python
light['백색광추정광량'] = light['백색광추정광량'].interpolate()
light['적색광추정광량'] = light['적색광추정광량'].interpolate()
light['청색광추정광량'] = light['청색광추정광량'].interpolate()
```

### 
백/적/청색광추정광량의 결측치가 모두 채워짐을 알 수 있다.

```python
light.isna().sum().to_frame(name='NaN')
```

### 
### 3. 총추정광량 결측치 보간

```python
light['총추정광량'] = np.where(light['총추정광량'].isna(),
                          light['백색광추정광량']+light['적색광추정광량']+light['청색광추정광량'], light['총추정광량'])
light.isna().sum().to_frame(name='NaN')
```

### 
모든 결측치가 채워짐을 알 수 있다.

```python
fig = make_subplots(rows=4, cols=1, shared_xaxes=True)

for i, c in enumerate(light.columns, start=1):
    fig.add_trace(go.Scatter(x=df['시간'], y=light[c]), row=i, col=1)
custom_legend_name(light.columns)
fig.show()
```
