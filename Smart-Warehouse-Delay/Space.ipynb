#
### **🏗️ 13. 창고 구조 / 공간**
* storage_density_pct → 저장 밀도
* vertical_utilization → 수직 공간 활용률
* racking_height_avg_m → 평균 랙 높이
* cross_dock_ratio → 크로스도킹 비율

```python
space = df[["storage_density_pct","vertical_utilization",
        "racking_height_avg_m","cross_dock_ratio"]]
space
```

```python
missing = space.isnull().sum()

colors = sns.color_palette("rocket", len(missing))

plt.figure(figsize=(12, 5))
missing.plot(kind='bar', color=colors)

plt.title('Missing Values - space')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

#
### **상관관계 분석**
* 상관관계가 없는 것으로 판단

```python
corr = space.corr()
corr
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='Blues', fmt=".2f")

plt.title('Correlation - Space Group')
plt.xticks(rotation=45)
plt.show()
```
